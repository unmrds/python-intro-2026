# Managing Flow 1: For Loops
----------------------------

It is often the case in programming that we need to perform the same operation on multiple objects or elements. Python and other programming languages have a defined process for this - loops.

There are two types of loops in Python, ```for``` loops and ```while``` loops. Today we are only going to look at ```for``` loops.

````python
# Start with a list - we often use for loops to iterate over the elements in a list
guitars = ['seagull', 'epiphone', 'guild', 'charvel']
````

````python
# we can print one at a time
print(guitars[0])
print(guitars[1])
````

````python
# and etc.
# But a more efficient way to do this is with a loop
for guitar in guitars:
    print(guitar)
````

````python
"""
What is happening above? For loop syntax:

for loop_variable in iterable:
    do_something

Note the loop_variable name can be anything, but as with other variable names it should make sense.
The body of the loop - code executed as part of the loop - is indented. 

An "iterable" is any object that can be iterated over - we'll see examples, but lists are commonly used as iterables.
The process exits the loop after the last item in the iterable has been operated upon.
"""
````

````python
# loop variable demonstration
for g in guitars:
    print(g)
````

````python
for supernova in guitars:
    print(supernova)
````

````python
# both work, but do they make as much sense as singular 'guitar'? Always consider human readability of your code.
# as an example of "do something," let's capitalize the name brands of our guitars
# there are already functions for this in Python that we can incorporate into our list

# Note that except for the use of str.capitalize() function, this loop only uses things we have already covered!
# comparing the code with the output, what does '\n' do?
print("List elements before the loop:", guitars)
for guitar in guitars:
    print("\nBegin the loop for:", guitar)
    i_pos = guitars.index(guitar)
    print("The index position of this element is:", i_pos)
    upper_cased = guitar.capitalize()
    print("Corrected brand name:", upper_cased)
    guitars[i_pos] = upper_cased

print("\nList elements after the loop:", guitars)
````

````python
# While loops
# Only as time permits

# For loops repeat the loop actions until the last element in the iterable has been processed.
# This is generally true - we can pass and exit loops but aren't going to cover that today.

# Alternatively, we can use a while loop that repeats actions only until a certain condition is met.
n = 1
print("Before while loop.\n")
while n <= 10:
    print("In the while loop. The current value of n is:", n)
    n = n + 1
print("\nAfter the while loop. The final value of n is:", n)
````
