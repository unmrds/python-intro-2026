```python
# 0 is equivalent to False
# make sure the output isn't a gotcha - the condition being evaluated is whether 0 is equivalent to False
# since that is the case in Python, the following evaluates to True
print(0 == False)
```

```python
# this seems awfully dry - why go into data types now?
# because often errors can result from trying to perform operations on objects of different data types
# this works - two args
print(me, n)
```
