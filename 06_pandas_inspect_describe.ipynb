```python
# the trucated output makes it hard to tell if other variables are affected by the inconsistent capitalization of T/F
# how can we dig deeper - it's important to be thorough!
# there are multiple ways, but one is to use describe()

# by default, outputs desc stats of numeric dtypes
census.describe()
```

```python
# change the default behavior of describe()
# all data types
census.describe(include='all')
```

```python
# or by data types
# we have seen numeric types
# 'unique' doesn't count NAs, so we know there are instance of True in each boolean variable
census.describe(include="boolean")
```

```python
# object or char data type
# this is useful - count of unique is less than the total count
# what does this mean? were some squirrels counted twice? were different squirrels given the same unique_squirrel_id?
# 5 squirrels does not seem statistically significant, but for transparency/QA/QC we should deal with this prior to any analysis
census.describe(include='object')
```

```python
# So - above_ground_sighter has 41 unique values, so it can't be changed to boolean
# what are the other values?
# group the data and see - we will come back to this later
census.groupby('above_ground_sighter')['unique_squirrel_id'].count()
```

```python
"""
above_ground_sighter variable definition:

"Measurement. For squirrel sightings on the ground plane, fields were populated with a value of “FALSE.”"

Do we assume the numeric entries represent feet above the ground plane? If so, how is FALSE different from 0?
More importantly, how do we use these data? If for example we wanted to know if certain species climb higher or are above the ground plane
more frequently?

"""
# we can convert to numeric data, but all FALSE values will be converted to NaN
# if we are interested in using these data, this may be the only option
print(pd.to_numeric(census['above_ground_sighter'], errors='coerce'))
```

```python
# but we don't want to overwrite existing values - it's better to keep the original data intact
# and pandas doesn't like to overwrite cell values in any case
# so we create a new column, but not on inspection that we lost a lot of data in this col
#     - compare with the original above_ground_sighter values
census['above_ground_sighter_n'] = pd.to_numeric(census['above_ground_sighter'], errors='coerce')
census.info()
```

```python
# before further inspection or analysis, 'date' is a dtype in pandas
# we need to convert the date data 
# pandas has excellent datetime handling
census['iso_date'] = pd.to_datetime(census['date'], format='%m%d%Y')
census.info()
```

```python
# What if we only want certain cols
# same syntax - [expression, expression] where rows and cols are selected based on whether the expression
# evaluates to True for that row/column - we can demonstrate as needed
census.loc[census['hectare']=='11H', ['iso_date', 'shift', 'hectare_squirrel_number']]
```
