# Managing Flow 2: Conditional Logic
------------------------------------

In addition to repeating actions, within a computational workflow we may often want to execute code conditionally. That is, whether or not some action is taken depends on characteristics of the object or variable we are operating upon.

We do this using conditional branching.

````python
# We are evaluating the truth of a condition
# Redo a few times with different values.
n = 5
if n < 10:
    print("n is less than 10.")
````

````python
# what's going on here?
# Boolean data types in action!
# Redo with different values and operators.
n = 5
print(n < 10)
````

````python
# our original example only handles cases where the condition evaluates to True
# use 'else' to handle ALL other cases. Note that this is optional - we don't have to do anything with the other cases.

# 'else' does not require a condition - it handles all cases that don't satisfy an earlier condition
n = 5
if n < 10:
    print("n is less than 10")
else:
    print("n is equal to or greater than 10")
````

````python
# In the above example, we are handling two cases with the 'else' branch
#     - n is greater than 10
#     - n is equal to 10
# This could be okay depending on what we want to do, but we can handle multiple other cases with 'elif' (else if)
n = 5
if n < 10:
    print("n is less than 10")
elif n == 10:
    print('n is equal to 10')
else:
    print('n is greater than 10')
````

````python
# Use this with some of our other concepts so far
# Sorting odd and even numbers
# Two new things -> range() and modulo division

# Modulo division - remainder division
print("No remainder (result is an integer):", 6/2)
print("Remainder (result is a decimal):", 6/4)
````

````python
# Modulo division returns the remainder
print("No remainder (6 is evenly divisible by 2):", 6%2)
print("Remainder (6 is not evenly divisible by 4, there is a remainder of 2):", 6%4)
````

````python
# We'll come back to this in a moment!
# Using range(start: stop: increment)
one_ten = range(1, 10)
print(type(one_ten))
````

````python
for n in one_ten:
  print(n)
````

````python
# can specify an increment
by_threes = range(0, 30, 3)
for t in by_threes:
    print(t)
````

````python
# Now we can create a range of numbers of any size and sort them into
# odds and evens

# after first example, redo with different ranges and increments - note that added complexity doesn't affect the code
odds = []
evens = []
for i in range(1, 100):
    # even numbers will be evenly divisible by 2, so use that as our truth test
    if i%2 == 0:
        evens.append(i)
    else:
        odds.append(i)

print("List of even numbers:", evens)
print("List of odd numbers:", odds)
````

````python
# It is possible to have multiple conditions in one truth test
# and - both conditions have to evaluate to True
n = 4
print(n > 1 and n < 6)
````

````python
# or - only one condition has to evaluate to True
# redo with 'and' to demonstrate
print(n < 1 or n > 2)
````

````python
# letter grades example
scores = [99, 92, 77, 95, 90, 90, 86]
for s in scores:
    print("\nDetermining letter grade for:", s)
    if s > 96 and s <=100:
        letter_grade = "A+"
        print("The score", s, "represents a letter grade of", letter_grade)
    elif s >= 93 and s <= 95:
        letter_grade = "A"
        print("The score", s, "represents a letter grade of", letter_grade)
    elif s >= 90 and s < 93:
        letter_grade = "A-"
        print("The score", s, "represents a letter grade of", letter_grade)
    else:
        print("The score", s, "is below 90.")
````
